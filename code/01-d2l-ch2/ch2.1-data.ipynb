数据访问和写入

```python
X[-1],X[1:3]#索引和切片 从后往前数，从后往前数的索引从 -1 开始

X[1,2] = 9#通过索引访问和修改元素
print(X)    
X[0:2,:]=12#通过切片索引访问和修改元素,:表示所有列,0:2表示前两行,1:3表示第二行,
print(X)
```

运行一操作可能会导致为新结果分配内存，所以学习原地操作

```python
before=id(X) # 记录X的内存地址
Y=Y+X
id(Y) == before # False, Y的内存地址发生了变化  
```

```python
Z=torch.zeros_like(Y)#创建一个形状和Y相同的全零张量
print(id(Z) == id(Y),id(Z)) # False, Z和Y的内存地址不同,Z的内存地址是新的
Z[:]=Y+X
print('id(Z):',id(Z))# True, Z的内存地址没有变化
```
